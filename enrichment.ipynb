## Дополнение рандомными значениями

```python
sample = auto_ex02.sample(n=200, random_state=21)
refund_values = auto_ex02['Refund'].dropna().unique()
fines_values = auto_ex02['Fines'].dropna().unique()
sample['Refund'] = np.random.choice(refund_values, size=200)
sample['Fines'] = np.random.choice(fines_values, size=200)
concat_rows = pd.concat([auto_ex02, sample], ignore_index=True)
```

```python
years = np.random.randint(1980, 2020, size=len(concat_rows))
year_series = pd.Series(years, name='Year')
concat_rows['Year'] = year_series
fines = concat_rows
```

```python
with open("surname.json", "r") as f:
    surnames = [x.split(",")[0].strip("[").strip("\"") for x in f.readlines()[1:]]
print(surnames)
```

```python
num_cars = 100  # например, 100 машин
car_numbers = [f"ABC{str(i).zfill(3)}" for i in range(1, num_cars + 1)]

# Создаем случайные фамилии без спецсимволов, равное количеству уникальных номеров
surnames_unique = np.random.choice(surnames, size=num_cars, replace=False)

# Создаем DataFrame
owners = pd.DataFrame({
    'CarNumber': car_numbers,
    'SURNAME': surnames_unique
})
```

```python
owners
```

```python
# Создаем новые строки
new_rows = pd.DataFrame({
    'CarNumber': ['QW12345', 'ZX67890', 'ER23456', 'TY78901', 'UI34567'],
    'Refund': [1.5, 2.0, 0.5, 3.0, 1.0],
    'Fines': [1500, 3000, 500, 2000, 1000],
    'Make': ['Honda', 'Nissan', 'Mazda', 'Hyundai', 'Kia'],
    'Model': ['Civic', 'Sentra', '3', 'Elantra', 'Soul'],
    'Year': [2007, 2006, 2005, 1998, 2000]
})

# Объединяем с concat_rows
fines = pd.concat([fines, new_rows], ignore_index=True)
```

```python
fines
```

```python
owners = owners[:-20]

# 2. Создаем три новых наблюдения
new_owners = pd.DataFrame({
    'SURNAME': ['Ivan', 'Anna', 'Sergey'],
    'CarNumber': ['AB12345', 'CD67890', 'EF23456']
    # Добавьте другие колонки, если есть, и заполните их
})

# 3. Добавляем новые наблюдения в owners
owners = pd.concat([owners, new_owners], ignore_index=True)
```

```python
# Группируем по марке, модели и году, считаем сумму штрафов
grouped = fines.groupby(['Make', 'Model', 'Year'])['Fines'].sum().reset_index()

# Создаем таблицу с уровнем модели и маркой, а годы как колонки
pivot_table = grouped.pivot_table(
    index=['Make', 'Model'],
    columns='Year',
    values='Fines',
    aggfunc='sum'
)
```
