## Расчет времени и добавление колонок разными методами

```python
def calc_for_loop(df):
    result = []
    for i in range(len(df)):
        value = df.iloc[i]['Fines'] / df.iloc[i]['Refund'] * df.iloc[i]['Year']
        result.append(value)
    return result
```

```python
%%timeit
df['calculated_for_loop'] = calc_for_loop(df)
```

```python
df
```

```python
def calc_iterrows(df):
    result = []
    for index, row in df.iterrows():
        value = row['Fines'] / row['Refund'] * row['Year']
        result.append(value)
    return result
```

```python
%%timeit
df['calculated_iterrows'] = calc_iterrows(df)
```

```python
df
```

```python
%%timeit
df['calculated_apply'] = df.apply(lambda row: row['Fines'] / row['Refund'] * row['Year'], axis=1)
```

```python
df
```

```python
%%timeit
fines_series = df['Fines']
refund_series = df['Refund']
year_series = df['Year']
df['calculated_series'] = (fines_series / refund_series) * year_series
```

```python
df
```

```python
%%timeit
values_fines = df['Fines'].values
values_refund = df['Refund'].values
values_year = df['Year'].values
df['calculated_values'] = (values_fines / values_refund) * values_year
```
