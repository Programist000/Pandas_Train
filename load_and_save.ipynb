```python
import pandas as pd
```

## Импортировал модуль pandas, считывание csv-файла

```python
df = pd.read_csv(
    'feed-views.log',
    sep='\t',             # разделитель табуляция
    skiprows=[2, 3],      # пропускаем 3-ю и 4-ю строки (индексы 2 и 3)
    skipfooter=2,         # пропускаем последние две строки
    names=['datetime', 'user'],  # задаем имена колонок
    index_col='datetime',        # делаем datetime индексом
    engine='python'
)
# Переименование колонки index
df.index.name = 'date_time'

# Сохранение с разделителем ';'
df.to_csv('feed-views-semicolon.log', sep=';')
df
```
