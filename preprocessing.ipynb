## Импорт + считывание и заданеие индексов

```python
cars = pd.read_csv("auto.csv")
```

```python
# Предположим, есть столбец 'user' или 'ID'
cars = cars.set_index('ID')  # заменит текущий индекс на столбец 'user'
```

```python
cars
```

```python
print(cars.count())
```

## Удаление дубликатов и проверки

```python
# Удаляем дубликаты, учитывая только указанные столбцы, оставляя последние
cars = cars.drop_duplicates(subset=['CarNumber', 'Make_n_model', 'Fines'], keep='last')
```

```python

# Проверяем число наблюдений
print(f"Количество наблюдений после удаления дубликатов: {cars.count()}")
```

```python
cars
```

```python
print(cars.isnull().sum())
```

```python
# Количество строк в DataFrame
n_rows = len(cars)

# Минимальное количество непустых значений для сохранения столбца
min_non_nulls = n_rows - 500

# Удаляем столбцы с более чем 500 пропущенными значениями
cars = cars.dropna(axis=1, thresh=min_non_nulls)
```

```python
cars
```

```python
print(cars.isnull().sum())
```

```python
cars['Refund'] = cars['Refund'].ffill()
```

```python
print(cars.isnull().sum())
```

```python
cars
```

```python
mean_fines = cars['Fines'].mean()
cars['Fines'] = cars['Fines'].fillna(mean_fines)
```

```python
print(cars.isnull().sum())
```

```python
cars
```

```python
def split_make_model(x):
    parts = x.split(' ', 1)
    if len(parts) == 2:
        return parts
    else:
        return [parts[0], '']

cars['Make'], cars['Model'] = zip(*cars['Make_n_model'].apply(split_make_model))
```

```python
cars
```

```python
# 2. Удаляем исходный столбец
cars = cars.drop(columns=['Make_n_model'])
```

```python
cars
```

```python
cars.to_json('auto.json', orient='records')
```
