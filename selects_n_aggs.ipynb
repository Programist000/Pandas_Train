```python
auto_ex02 = auto_ex02.set_index('CarNumber')
```

## Выводы по проверкам

```python
fines_more_2100 = auto_ex02[auto_ex02['Fines'] > 2100]
```

```python
fines_more_2100_and_refound_equals_2 = fines_more_2100[fines_more_2100["Refund"] == 2]
```

```python
print("fines are greater than 2100:")
print(fines_more_2100)
```

```python
print("fines are greater than 2100 and refunds are equal 2:")
print(fines_more_2100_and_refound_equals_2)
```

```python
print("The rows where the models are from the list: ['Focus', 'Corolla']:")

# Используем .isin() для проверки, входит ли значение в список
print(auto_ex02[auto_ex02["Model"].isin(["Focus", "Corolla"])])
```

```python
print("The rows where the car number is from the list: ['Y7689C197RUS', '92928M178RUS', '7788KT197RUS', 'H115YO163RUS', 'X758HY197RUS']")

car_numbers = ['Y7689C197RUS', '92928M178RUS', '7788KT197RUS', 'H115YO163RUS', 'X758HY197RUS']
print(auto_ex02.loc[auto_ex02.index.isin(car_numbers)])
```

## Группировка

```python
auto_ex02.groupby('Make')['Fines'].median()
```

```python
auto_ex02.groupby(['Make', 'Model'])['Fines'].median()
```

```python
auto_ex02.groupby(['Make', 'Model'])['Fines'].count()
```

```python
auto_ex02.groupby(['Make', 'Model'])['Fines'].min()
auto_ex02.groupby(['Make', 'Model'])['Fines'].max()
```

```python
auto_ex02.groupby(['Make', 'Model'])['Fines'].std()
```

```python
fines_counts = auto_ex02.index.value_counts().sort_values(ascending=False)
print(fines_counts)
```

```python
sum_fines_by_car = auto_ex02.groupby(auto_ex02.index)['Fines'].sum().sort_values(ascending=False)
print(sum_fines_by_car)
```

```python
top_car_by_fines = sum_fines_by_car.index[0]
top_rows_by_fines = auto_ex02.loc[top_car_by_fines]
print(top_rows_by_fines)
```

```python
model_counts = auto_ex02.groupby(auto_ex02.index)['Model'].nunique()
multiple_models = model_counts[model_counts > 1]
print(multiple_models)
```

```python
models = ['Focus', 'Corolla']
auto_ex02.loc[auto_ex02['Model'].isin(models)]
```
